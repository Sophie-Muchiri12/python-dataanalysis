# NumPy 

NumPy (or Numpy) is a Linear Algebra Library for Python, the reason it is so important for Data Science with Python is that almost all of the libraries in the PyData Ecosystem rely on NumPy as one of their main building blocks.

Numpy is also incredibly fast, as it has bindings to C libraries. For more info on why you would want to use Arrays instead of lists, check out this great [StackOverflow post](http://stackoverflow.com/questions/993984/why-numpy-instead-of-python-lists).

We will only learn the basics of NumPy, to get started we need to install it!

# Numpy Arrays

NumPy arrays are the main way we will use Numpy throughout the course. Numpy arrays essentially come in two flavors: vectors and matrices. Vectors are strictly 1-d arrays and matrices are 2-d (but you should note a matrix can still have only one row or one column).

Let's begin our introduction by exploring how to create NumPy arrays.

## Creating NumPy Arrays

### From a Python List

We can create an array by directly converting a list or list of lists:

```python
my_list = [1,2,3]
my_list
```

np.array(list) -> converts a python list into a numpy array

Numpy arrays support efficient numerical operations that regular python lists dont

example:
my_list = [1,2,3] ---> print(my_list*2) ---> this duplicates the list = [1,2,3,1,2,3]

But with numpy arrays:

arr = np.array([1, 2, 3])

print(arr * 2) ---> multiplies each element with 2
# [2 4 6]

```python
np.array(my_list)
```

```python
my_matrix = [[1,2,3],[4,5,6],[7,8,9]] # 3 rows and 3 columns
my_matrix
```

```python
np.array(my_matrix)
```

## Built-in Methods

There are lots of built-in ways to generate Arrays

### arange

Return evenly spaced values within a given interval.

```python
np.arange(0,10)
#arange produces range of numbers ([start],[stop],[step]) upto but not including the stop value
```

```python
np.arange(0,11,2) #jump in steps of 2
```

### zeros and ones

Generate arrays of zeros or ones

```python
np.zeros(3) #Why are the zeros written as 0. instead of 0?

# By default, NumPy creates a floating-point array:
```

```python
np.zeros((5,5))
```

```python
np.ones(3)
```

```python
np.ones((3,3))
```

### linspace
Return evenly spaced numbers over a specified interval.

```python
np.linspace(0,10,3)
```

```python
np.linspace(0,10,50)
```

## eye

Creates an identity matrix

An idenity matrix is a square matrix with 

1s on the main diagonal (top left to bottom right)

```python
np.eye(4)
```

## Random 

Numpy also has lots of ways to create random number arrays:

### rand
Create an array of the given shape and populate it with
random samples from a uniform distribution
over ``[0, 1)``.

```python
np.random.rand(2) # generates 2 numbers between 0 and 1
```

```python
np.random.rand(10) #generates 10 numbers between 0 and 1 
```

```python
np.random.rand(5,5) # generates a 5 x 5 matrix of numbers between 0 and 1
```

### randn

Return a sample (or samples) from the "standard normal" distribution. Unlike rand which is uniform:

```python
np.random.randn(10)
```

```python
np.random.randn(10,10)
```

```python
np.random.randn(5,5)
```

### randint
Return random integers from `low` (inclusive) to `high` (exclusive).

```python
np.random.randint(1,100) #generates a whole number between 1 and 100 but not including 100
```

```python
np.random.randint(1,100,10) #generates 10 random numbers between 1 and 100
```

## Reshape
Returns an array containing the same data with a new shape.

Reshape means ---> “Take the same numbers, but arrange them differently."

but not randomly once re-arranged however many times you try running it it will be the same

```python
arr.reshape(5,5)
```

### max,min,argmax,argmin

These are useful methods for finding max or min values. Or to find their index locations using argmin or argmax

```python
ranarr
```

```python
ranarr.max() #returns the largest values in the array
```

```python
ranarr.argmax() #returns the position(index) of the largest value
```

```python
ranarr.min() #returns the smallest values in the array
```

```python
ranarr.argmin() #returns the position(index) of the largest value
```

## Shape

Shape is an attribute that arrays have (not a method):

It tells you  the size/dimensions of a Numpy array basically how many rows,columns etc.

```python
# Vector
arr.shape # returns (25,) which means 1 row of data and 25 elements
```

```python
arr1 = np.array([[1, 2, 3],
                [4, 5, 6]])

print(arr1.shape) # 
```

```python
# Notice the two sets of brackets
arr.reshape(1,25)
```

```python
arr.reshape(1,25).shape
```

```python
arr.reshape(25,1)
```

```python
arr.reshape(25,1).shape
```
