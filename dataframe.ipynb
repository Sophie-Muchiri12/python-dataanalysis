```python
people = {
    "first":["corey","sophie", "cristian"],
    "last":["schafer","muthoni","rodriguez"],
    "email":["corey@gmail.com","muthoni@gmail.com","rodriguez@gmail.com"]
}
```

```python
#what is a dataframe?
#A DataFrame is a 2-dimensional labeled data structure with Rows(indexed) and 
#Columns(named)
```

```python
# creating a dataframe from the people object
import pandas as pd
```

```python
df = pd.DataFrame(people)
```
