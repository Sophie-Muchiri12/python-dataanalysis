```python
people = {
    "first":["corey","sophie", "cristian"],
    "last":["schafer","muthoni","rodriguez"],
    "email":["corey@gmail.com","muthoni@gmail.com","rodriguez@gmail.com"]
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
#sorting by lastname
```

```python
df.sort_values(by='last')
```

```python
#in descending order
df.sort_values(by='last', ascending=False)
```

```python
#sorting  columns with similar values

df.sort_values(by=['last','first'], ascending=False)
```
