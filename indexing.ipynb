```python
people = {
    "first":["corey","sophie", "cristian"],
    "last":["schafer","muthoni","rodriguez"],
    "email":["corey@gmail.com","muthoni@gmail.com","rodriguez@gmail.com"]
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
#What is indexing in pandas
df.set_index('email')
```

```python
df
#index doesnt change the dataframe unless we explicitly tell pandas to do so
```

```python
df.set_index('email', inplace=True)
```

```python
df
```

```python
df.index
```

```python
#why is this useful (indexing)
#index becomes label when we use loc[]
df.loc['corey@gmail.com']
```

```python
df.loc['corey@gmail.com','last']
```
