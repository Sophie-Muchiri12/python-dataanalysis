```python
import pandas as pd
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
weeks_data =[1,2,3,4,5]
sales_data = [2,4,5,4,5]
```

```python
df = pd.DataFrame({
    "Weeks":weeks_data,
    "Sales":sales_data
})
```

```python
df
```

```python
#at the x axis we have the input which in this case are the weeks so:
X = df['Weeks'].values
X
```

```python
Y = df['Sales'].values
Y
```

### so in the above we've separated our features and target as X and Y respectively

#### Features are simply:

#### The input variables (the information you use to make a prediction)



```python
#training vs testing
```

```python
#we look for mean of x and y
mean_x = np.mean(X)
mean_y = np.mean(Y)
```

```python
n = len(X)

numer = 0
denom = 0

for i in range(n):
    numer += (X[i] - mean_x) * (Y[i] - mean_y)
    denom += (X[i] - mean_x) ** 2

m = numer/denom

# c = y' - mx'

c = mean_y - (m * mean_x)

print(f"m = {m}")
print(f"c = {c}")
    
```

```python
#plotting
max_x = np.max(X) + 1 
min_x = np.min(X) - 1
# x axis will start from 0 to 6
#linspace is used for spacing
x = np.linspace(min_x,max_x)
y = (m * x) + c

plt.plot(x,y,color='red', label='Regression Line', linestyle='--')
plt.scatter(X,Y,color='blue',label='Actual Data point')
plt.xlabel('Weeks')
plt.ylabel('Sales')
plt.title('Linear Regression Demo')
plt.grid(True)
plt.legend(loc='best')
```
