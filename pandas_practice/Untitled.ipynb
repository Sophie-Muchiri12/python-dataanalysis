# The difference between the len() and .sum() is that: sum will only count the True value in a boolean mask while len will count both True and False values you'll have to us in different scenarios:

for example if we use .sum directly on the boolean mask thats fine but you cant do the same with len

the best case  for len is after filtering so meaning putting the boolean mask inside the dataframe to filter it

like so:

# len(df[(df['Marks'] >= 90) & (df['Marks'] <= 100)])
