# 2. Change the performance rating to "Outstanding" for employees earning more than 90,000 and currently rated "Excellent".

```python
employees.loc[(employees['salary'] >= 90000) & (employees['performance'] == 'Excellent'), 'performance'] = employees.loc[(employees['salary'] >= 90000) & (employees['performance'] == 'Excellent'), 'performance'] = 'Outstanding' 
```
