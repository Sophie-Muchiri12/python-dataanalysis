# Cleaning Data
## How to handle missing values and Casting Data types

```python
import pandas as pd
import numpy as np
```

```python

people = {
    "first": ["corey", "sophie", "cristian", "john", "jane", np.nan, None, 'NA', 'Anonymous'],
    "last": ["schafer", "muthoni", "rodriguez", "NA", "Doe", np.nan, np.nan, 'Missing', 'Unknown'],
    "email": ["corey@gmail.com", "muthoni@gmail.com", "Anonymous@email.com",
              None, "janeDoe@gmail.com", None, np.nan, 'Anonymous@email.com', 'NA'],
    "age": ['33', '55', '63', '36', '46', None, None, 'Missing', None]
}
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df.dropna()#using default arguments
```

```python
df.dropna(axis='index',how='all')#using default arguments
```

### index and columns arguments
#### Setting to index meaning drop rows with missing values
### how/criteria to drop rows or columns if set to any means drop any row with missing values can be set to all meaning only drop the row if all columns have missing data

```python
df.dropna(axis='columns',how='any')
```

```python
df.dropna(axis='index',how='any',subset=['last','email'])
```

```python
#custom missing values
df.replace('NA',np.nan,inplace=True)
df.replace('Missing',np.nan,inplace=True)
```

```python
df.isna()
```

```python
df.fillna('MISSING')
```
