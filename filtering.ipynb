```python
people = {
    "first":["corey","sophie", "cristian","john","jane"],
    "last":["schafer","muthoni","rodriguez","Doe","Doe"],
    "email":["corey@gmail.com","muthoni@gmail.com","rodriguez@gmail.com",
             "johnDoe@gmail.com","janeDoe@gmail.com"]
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
filt_last = (df['last'] == 'Doe') #can put parenthesis just clean code
```

```python
filt_last
```

```python
df[filt_last]
```

```python
df.loc[filt_last]
```

```python
df.loc[filt_last, 'email']
```

```python
df.loc[filtor,'email']
```

```python
filtand = (df['last'] == 'Doe') & (df['first'] == 'jane')
```

```python
df.loc[filtand,'email']
```
