## Why Transpose a DataFrame?

Transposing a DataFrame swaps its rows and columns.

```python
df.T
```

### 1. Change the Orientation for Analysis

Sometimes your data is easier to analyze when observations become columns and variables become rows.

Before:

| Name  | Math | Science |
| ----- | ---- | ------- |
| Alice | 85   | 90      |
| Bob   | 78   | 88      |

After
|         | Alice | Bob |
| ------- | ----- | --- |
| Math    | 85    | 78  |
| Science | 90    | 88  |
