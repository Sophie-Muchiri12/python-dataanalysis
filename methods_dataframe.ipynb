## Why `df.describe()` Shows This Output for Your Data

Your DataFrame contains **text (string) columns**, so `describe()` gives **categorical statistics instead of numeric ones**.

---

## What Your Output Means

| Statistic | Meaning |
|------------|----------|
| **count** | Number of non-missing values |
| **unique** | Number of distinct values |
| **top** | Most frequent value (mode) |
| **freq** | How many times the most common value appears |

## why would we want to sort by multiple columns:
### 1. Better grouping + ordering (Department → Salary)

### 2. Makes patterns visible eg: Students in same class ordered by performance

### 3. Essential for ranking within groups

### NB: sort_values always sort in ascending order unless you explicitly set ascending to false -> ascending=False

```python
finance_df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Helen"],
    "department": ["HR", "IT", "IT", "HR", "Finance", "Finance", "IT", "HR"],
    "salary": [50000, 80000, 75000, 52000, 90000, 87000, 78000, 51000],
    "age": [25, 30, 28, 35, 40, 38, 27, 29]
})
```

```python
finance_df
```

```python
#sort by department in ascending order but salary descending order we look for the highest to lowest paid in all departments
finance_df.sort_values(['department','salary'],ascending=[True,False])
```
